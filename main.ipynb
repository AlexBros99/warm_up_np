```python
rng= np.random.default_rng(seed=42)
```

```python
rng.random((3,4))
```

```python
rng.permutation(10)
```

```python
plt.hist(rng.random(100000),density=True,bins=100,histtype="step", color="blue", label="rand")
plt.hist(rng.standard_normal(100000),density=True,bins=100,histtype="step",color="red", label="randn")
plt.axis([-2.5, 2.5, 0, 1.1])
plt.legend(loc = "upper left")
plt.title("Random distributions")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```
